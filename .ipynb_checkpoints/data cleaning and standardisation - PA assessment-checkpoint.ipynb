```python
# Check number of rows with each value in "Transaction status"
import pandas as pd

# Load the CSV file
df = pd.read_csv("Transaction data stage 2.csv")

# Count occurrences of each unique value in the 'Transaction status' column
status_counts = df["Transaction status"].value_counts()

# Print the results
for status, count in status_counts.items():
    print(f"{status}: {count}")
```

```python
# Clean data - fixing typos, removing spaces, removing duplicates, standardising format, value type standardisation, etc)
import pandas as pd

def clean_transaction_data(file_path, output_path, removed_output_path):
    # Load the dataset
    df = pd.read_csv(file_path)
    
    # Standardize column names (fix typos, remove spaces, convert to title case)
    df.rename(columns={
        'sales channgel': 'Sales Channel'
    }, inplace=True)
    
    # Remove duplicates based on Transaction ID
    df = df.drop_duplicates(subset=['Transaction ID'])
    
    # Handle missing values
    df['Payment Method'].fillna('Unknown', inplace=True)
    
    # Ensure Amount (£) is numeric and remove rows with non-numeric values
    df['Amount (£)'] = pd.to_numeric(df['Amount (£)'], errors='coerce')  # Convert to numeric, forcing errors to NaN
    removed_data = df[df['Amount (£)'].isna()]  # Capture removed data
    df = df.dropna(subset=['Amount (£)'])  # Drop rows with NaN in Amount (£)
    
    # Fill missing numeric values with the median after cleaning
    df['Amount (£)'].fillna(df['Amount (£)'].median(), inplace=True)
    
    # Convert Amount (£) to float
    df['Amount (£)'] = df['Amount (£)'].astype(float)
    
    # Standardize date format
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce').dt.date
    
    # Standardize categorical text values
    df['Payment Method'] = df['Payment Method'].str.strip().str.title()
    
    # Fix inconsistencies in Payment Method
    df['Payment Method'] = df['Payment Method'].replace({
        'creditcard': 'Credit Card', 
        'debitcard': 'Debit Card'
    })
    
    # Remove rows where Transaction status is 'Failed'
    removed_data = pd.concat([removed_data, df[df['Transaction status'] == 'Failed']])
    df = df[df['Transaction status'] != 'Failed']
    
    # Remove negative amounts
    removed_data = pd.concat([removed_data, df[df['Amount (£)'] < 0]])
    df = df[df['Amount (£)'] >= 0]
    
    # Handle outliers using IQR
    Q1 = df['Amount (£)'].quantile(0.25)
    Q3 = df['Amount (£)'].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df['Amount (£)'] < (Q1 - 1.5 * IQR)) | (df['Amount (£)'] > (Q3 + 1.5 * IQR))]
    removed_data = pd.concat([removed_data, outliers])
    df = df[(df['Amount (£)'] >= (Q1 - 1.5 * IQR)) & (df['Amount (£)'] <= (Q3 + 1.5 * IQR))]
    
    # Ensure correct data types
    df = df.astype({
        'Transaction ID': 'str',
        'Amount (£)': 'float',
        'Customer Segment': 'str'
    })
    
    # Save cleaned dataset
    df.to_csv(output_path, index=False)
    
    # Save removed data
    removed_data.to_csv(removed_output_path, index=False)
    
    print(f"Cleaned data saved to {output_path}")
    print(f"Removed data saved to {removed_output_path}")

# Example usage
input_file = 'Transaction data stage 2.csv'  # Update with your file path
output_file = 'cleaned_transactions.csv'
removed_output_file = 'removed_transaction_data.csv'
clean_transaction_data(input_file, output_file, removed_output_file)
```

```python
# Correct spelling errors in columns
import pandas as pd
from difflib import get_close_matches

def closest_match(value, valid_options):
    match = get_close_matches(value, valid_options, n=1, cutoff=0.7)
    return match[0] if match else value

def update_sales_channel_column(file_path):
    # Load the dataset
    df = pd.read_csv(file_path)
    
    # Rename the column
    df.rename(columns={'Sales channgel': 'Sales channel'}, inplace=True)
    
    # Define valid options
    valid_payment_methods = ["Credit Card", "Mobile Payment", "Digital Wallet", "Bank Transfer", "Debit Card"]
    valid_merchant_categories = ["Entertainment", "Food & Beverage", "Retail", "Services", "Travel"]
    valid_locations = ["London", "Manchester", "Birmingham", "Bristol"]
    valid_sales_channels = ["Online", "Mobile App", "Subscription", "Social Media", "In-Store"]
    valid_promotions = ["First-Time User Discount", "Loyalty Points Redeemed", "No", "Yes", "Seasonal Offer"]
    
    # Standardize column values
    df['Payment Method'] = df['Payment Method'].astype(str).apply(lambda x: closest_match(x, valid_payment_methods))
    df['Merchant Category'] = df['Merchant Category'].astype(str).apply(lambda x: closest_match(x, valid_merchant_categories))
    df['Location'] = df['Location'].astype(str).apply(lambda x: closest_match(x, valid_locations))
    df['Sales channel'] = df['Sales channel'].astype(str).apply(lambda x: closest_match(x, valid_sales_channels))
    df['Promotion/Discount Applied'] = df['Promotion/Discount Applied'].astype(str).apply(lambda x: closest_match(x, valid_promotions))
    
    # Save the updated CSV file
    df.to_csv(file_path, index=False)
    
    print(f"Updated column names and standardized values. Changes saved to {file_path}")

# Run the function
file_path = 'cleaned_transactions.csv'
update_sales_channel_column(file_path)
```

```python
import pandas as pd
from difflib import get_close_matches

def closest_match(value, valid_options):
    match = get_close_matches(value, valid_options, n=1, cutoff=0.7)
    return match[0] if match else value

def update_transaction_data(file_path, removed_file_path, new_file_path):
    # Load the dataset
    df = pd.read_csv(file_path)
    
    # Rename the column to title case
    df.rename(columns={'Sales channel': 'Sales Channel'}, inplace=True)
    
    # Remove rows where Transaction status is 'Chargeback'
    removed_data = df[df['Transaction status'] == 'Chargeback']
    df = df[df['Transaction status'] != 'Chargeback']
    
    # Load the second dataset
    df_stage2 = pd.read_csv('Transaction data stage 2.csv')
    
    # Extract rows where Transaction status is 'Failed'
    failed_transactions = df_stage2[df_stage2['Transaction status'] == 'Failed']
    
    # Extract rows where Transaction status is 'Chargeback'
    chargeback_transactions = df_stage2[df_stage2['Transaction status'] == 'Chargeback']
    
    # Load existing removed data
    try:
        existing_removed_data = pd.read_csv(removed_file_path)
        removed_data = pd.concat([existing_removed_data, removed_data, failed_transactions, chargeback_transactions])
    except FileNotFoundError:
        removed_data = pd.concat([removed_data, failed_transactions, chargeback_transactions])
    
    # Save the cleaned dataset
    df.to_csv(new_file_path, index=False)
    
    # Save the removed data without overwriting existing data
    removed_data.to_csv(removed_file_path, index=False)
    
    print(f"Updated column name to title case, removed 'Chargeback' transactions, added 'Failed' and 'Chargeback' transactions from stage 2. Changes saved to {new_file_path} and {removed_file_path}")

# Run the function
file_path = 'cleaned_transactions.csv'
removed_file_path = 'removed_transaction_data.csv'
new_file_path = 'cleaned_transactions_updated.csv'
update_transaction_data(file_path, removed_file_path, new_file_path)
```
