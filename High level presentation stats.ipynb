```python
# percentage of transactions by payment method

import pandas as pd

def calculate_payment_method_percentage(file_path: str):
    # Load the dataset
    df = pd.read_csv(file_path)
    
    # Check if 'Payment Method' column exists
    if 'Payment Method' not in df.columns:
        print("Error: 'Payment Method' column not found in the dataset.")
        return
    
    # Calculate value counts as percentages
    payment_method_counts = df['Payment Method'].value_counts(normalize=True) * 100
    
    # Print results
    print("Percentage of rows by Payment Method:")
    print(payment_method_counts.round(2).to_string())

if __name__ == "__main__":
    file_path = "cleaned_transactions.csv"  # Adjust if needed
    calculate_payment_method_percentage(file_path)
```

```python
# percentage of each customer segment that has 'No' in the Promotion/Discount Applied column

def calculate_no_discount_percentage(file_path: str):
    # Load the dataset
    df = pd.read_csv(file_path)
    
    # Check if required columns exist
    required_columns = {'Customer Segment', 'Promotion/Discount Applied'}
    if not required_columns.issubset(df.columns):
        print("Error: Required columns not found in the dataset.")
        return
    
    # Filter rows where 'Promotion/Discount Applied' is 'No'
    no_discount_df = df[df['Promotion/Discount Applied'] == 'No']
    
    # Calculate percentage of 'No' per customer segment
    no_discount_percentage = (no_discount_df['Customer Segment'].value_counts() / df['Customer Segment'].value_counts()) * 100
    
    # Print results
    print("Percentage of each Customer Segment with 'No' in Promotion/Discount Applied:")
    print(no_discount_percentage.round(2).to_string())

if __name__ == "__main__":
    file_path = "cleaned_transactions.csv"  # Adjust if needed
    calculate_no_discount_percentage(file_path)
```

```python
# what percentage of digital wallet and mobile payments under the 
# Payment Method columns have 'No' in the Promotion/Discount Applied column

def calculate_no_discount_for_digital_payments(file_path: str):
    # Load the dataset
    df = pd.read_csv(file_path)
    
    # Check if required columns exist
    required_columns = {'Payment Method', 'Promotion/Discount Applied'}
    if not required_columns.issubset(df.columns):
        print("Error: Required columns not found in the dataset.")
        return
    
    # Define digital payment methods
    digital_payments = {'Digital Wallet', 'Mobile Payment'}
    
    # Filter for digital payment transactions
    digital_df = df[df['Payment Method'].isin(digital_payments)]
    
    # Calculate percentage of 'No' in Promotion/Discount Applied
    no_discount_percentage = (digital_df['Promotion/Discount Applied'].eq('No').sum() / len(digital_df)) * 100
    
    # Print result
    print(f"Percentage of Digital Wallet and Mobile Payment transactions with 'No' in Promotion/Discount Applied: {no_discount_percentage:.2f}%")

if __name__ == "__main__":
    file_path = "cleaned_transactions.csv"  # Adjust if needed
    calculate_no_discount_for_digital_payments(file_path)
```

```python
# percentage of all transactions have 'No' in in the Promotion/Discount Applied column 

def calculate_no_discount_percentage(file_path: str):
    # Load the dataset
    df = pd.read_csv(file_path)
    
    # Check if the required column exists
    if 'Promotion/Discount Applied' not in df.columns:
        print("Error: 'Promotion/Discount Applied' column not found in the dataset.")
        return
    
    # Calculate percentage of transactions with 'No' in Promotion/Discount Applied
    no_discount_percentage = (df['Promotion/Discount Applied'].eq('No').sum() / len(df)) * 100
    
    # Print result
    print(f"Percentage of all transactions with 'No' in Promotion/Discount Applied: {no_discount_percentage:.2f}%")

if __name__ == "__main__":
    file_path = "cleaned_transactions.csv"  # Adjust if needed
    calculate_no_discount_percentage(file_path)
```
